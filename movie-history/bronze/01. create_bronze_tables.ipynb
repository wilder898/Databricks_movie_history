### Crear tablas para archivos CSV

#### Crear la tabla "movies"

```sql
DROP TABLE IF EXISTS movie_bronze.movies;
CREATE TABLE IF NOT EXISTS movie_bronze.movies(
    movield INT,
    title STRING,
    budget DOUBLE, 
    homePage STRING, 
    overview STRING, 
    popularity DOUBLE, 
    yearReleaseDate INT, 
    releaseDate DATE, 
    revenue DOUBLE, 
    durationTime INT, 
    movieStatus STRING, 
    tagline STRING, 
    voteAverage DOUBLE,
    voteCount INT
)
USING CSV
OPTIONS(path"abfss://bronze@stmoviesdatabricks.dfs.core.windows.net/movie.csv", header true);
```

```sql
%python
spark.read.option("header",True).csv("abfss://bronze@stmoviesdatabricks.dfs.core.windows.net/language.csv").display()
```

```sql
DROP TABLE IF EXISTS movie_bronze.languages;
CREATE TABLE IF NOT EXISTS movie_bronze.languages(
    languageId INT,
    languageCode STRING,
    languageName STRING
)
USING CSV
OPTIONS(path "abfss://bronze@stmoviesdatabricks.dfs.core.windows.net/language.csv", header true)
```

```sql
DROP TABLE IF EXISTS movie_bronze.genres;
CREATE TABLE IF NOT EXISTS movie_bronze.genres(
    genreId INT,
    genreName STRING
)
USING CSV
OPTIONS(path "abfss://bronze@stmoviesdatabricks.dfs.core.windows.net/genre.csv", header true)
```

```sql
SELECT * FROM movie_bronze.languages;
```
